```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
```

```python
df=pd.read_csv("../data/raw/smoking_prediction_etiquetado.csv")
```

```python
## Variable objetivo: representación
plt.figure(figsize=(6,4))
smoking_count = df.smoking.value_counts(normalize=True)*100
plt.pie(smoking_count, labels=["No Fumadores","Fumadores"], autopct="%1.1f%%", startangle=90)
plt.title("Porcentaje fumadores (smoking)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
gender_count = df.gender.value_counts()
gender_graph=sns.countplot(df,x="gender", hue="smoking")
gender_graph.bar_label(gender_graph.containers[0],fmt="%d", padding=0)
gender_graph.bar_label(gender_graph.containers[1],fmt="%d", padding=0)
plt.title("Género vs. Fumadores")
plt.xlabel("Género")
plt.ylabel("Candidad")
plt.ylim(0,20000)
plt.show()
```

```python
fig,axes=plt.subplots(1,2, figsize=(10,4))

tab_hleft=pd.crosstab(df['hearing(left)'], df['smoking'], normalize='index')[1]*100
tab_right=pd.crosstab(df['hearing(right)'], df['smoking'], normalize='index')[1]*100
barra_left=axes[0].bar(['0.04','0.08'],tab_hleft.values)
barra_right=axes[1].bar(['0.04','0.08'], tab_right.values)

axes[0].bar(['0.04','0.08'],tab_hleft.values, color=['green','red'])
axes[0].set_title("Audición izq. vs smoking")
axes[0].set_ylabel("% de fumadores")
axes[0].bar_label(barra_left, fmt="%.2f%%", padding=0)
axes[0].set_ylim(0,40)

axes[1].bar(['0.04','0.08'], tab_right.values, color=['green','red'])
axes[1].set_title("Audición der. vs smoking")
axes[1].set_ylabel("% de fumadores")
axes[1].bar_label(barra_right, fmt="%.2f%%")
axes[1].set_ylim(0,40)

plt.show()
```
