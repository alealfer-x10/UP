```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from xgboost import XGBClassifier
from xgboost import plot_importance
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.metrics import make_scorer, fbeta_score
from sklearn.metrics import classification_report

from sklearn.metrics import precision_recall_curve
```

```python
X_train, X_val, y_train, y_val = joblib.load("../data/processed/train_y_val.joblib")
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_val.shape
```

```python
y_val.shape
```

```python
%%time
clf = XGBClassifier(n_estimators =100, learning_rate=0.05, max_depth=5, random_state=42)
```

```python
%%time
clf.fit(X_train, y_train)
```

```python
%%time
train_pred = clf.predict(X_train)
val_pred = clf.predict(X_val)
```

```python
print("Train")
print(classification_report(y_train, train_pred))
print("--------------------------------------------------------")
print("Validation")
print(classification_report(y_val, val_pred))
```

```python
plt.figure(figsize=(10,8))
plot_importance(clf, max_num_features=12, importance_type="gain")
plt.show()
```

# Guardado de features y modelo no optimizado

```python
joblib.dump(clf, "../models/mod_XGB.joblib")
```

```python
joblib.dump(list(X_train.columns), "../models/features_XGB.joblib")
```

# Umbral de corte

```python
%%time
train_pred_proba_1 = clf.predict_proba(X_train)[:,1]
val_pred_proba_1 = clf.predict_proba(X_val)[:,1]
```

```python
train_pred_proba_1
```

```python
val_pred_proba_1
```

```python
plt.figure(figsize=(10,5))
sns.histplot(train_pred_proba_1, bins=100, kde=True, label="train", stat="density", alpha=0.7)
sns.histplot(val_pred_proba_1, bins=150, kde=True, label="validation", stat="density", alpha=0.7)
plt.axvline(0.5, color="red")
plt.title("Curva de probabilidades predichas fumadores train vs validation")
plt.xlabel("probabilidad")
plt.ylabel("density")
plt.legend()
plt.show()
```

```python
umbral=0.45642412
```

```python
y_train_pred_custom = (train_pred_proba_1 >= umbral).astype(int)
y_val_pred_custom = (val_pred_proba_1 >= umbral).astype(int)
```

```python
y_train_pred_custom
```

```python
y_val_pred_custom
```

```python
print("Train")
print(classification_report(y_train, y_train_pred_custom))
print("--------------------------------------------------------")
print("Validation")
print(classification_report(y_val, y_val_pred_custom))
```

```python
precision, recall, thresholds = precision_recall_curve(y_val, val_pred_proba_1)
```

```python
precision
```

```python
recall
```

```python
thresholds
```

```python
f1_score = 2 * (precision * recall) / (precision + recall)
```

```python
f1_score
```

```python
idx_mejor = np.argmax(f1_score[:-1])   # [:-1] porque precision/recall tienen un punto extra al final
mejor_umbral = thresholds[idx_mejor]
```

```python
print ("Mejor umbral: ",mejor_umbral)
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(thresholds, precision[:-1], label="Precision")
ax.plot(thresholds, recall[:-1], label="Recall")
ax.plot(thresholds, f1_score[:-1], label="F1-score")

ax.set_xlabel("Umbral de decisión")
ax.set_ylabel("Valor de la métrica")
ax.set_title("Precision, Recall y F1-score vs. umbral")
plt.axvline(0.45642412,label=f"Umbral óptimo = {mejor_umbral:.2f}" , color="red")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

# Optimización de hiperparámetros

```python
xgb_model = XGBClassifier(objective='binary:logistic', eval_metric = "logloss", random_state=42)
```

```python
xgb_model
```

```python
param_dist = {
    "n_estimators": [50, 100, 150, 200],
    "max_depth": [3,5,7],
    "learning_rate": [0.03, 0.04, 0.05],
    "subsample": [0.7, 0.8, 0.9]
}

"""
param_dist = {
    "n_estimators": [100, 200, 300, 400],
    "max_depth": [1,3,5],
    "learning_rate": [0.03, 0.04, 0.05],
    "subsample": [0.7, 0.8, 0.9]
}


param_dist = {
    "n_estimators": [400, 450, 500, 550],
    "max_depth": [1,3,5],
    "learning_rate": [0.02, 0.03, 0.05],
    "subsample": [0.7, 0.8, 0.9]
}


param_dist = {
    "n_estimators": [50, 100, 200, 500],
    "max_depth": [3,5,7],
    "learning_rate": [0.01, 0.05, 0.001],
    "subsample": [0.6, 0.8, 1]
}
"""
```

```python
fbeta_scorer = make_scorer(fbeta_score, beta=2)
```

```python
model_search = RandomizedSearchCV(
    estimator = xgb_model,
    param_distributions=param_dist,
    n_iter = 30,
    scoring = fbeta_scorer,
    cv = 5,
    random_state=42,
    verbose=2,
    n_jobs=-1
)
```

```python
model_search
```

```python
%%time
model_search.fit(X_train, y_train)
```

```python
print(model_search.best_params_)
```

```python
best_model = model_search.best_estimator_
```

```python
best_model
```

```python
%%time
best_model.fit(X_train, y_train)
```

```python
print("Train")
print(classification_report(y_train, train_pred))
print("--------------------------------------------------------")
print("Validation")
print(classification_report(y_val, val_pred))
```
